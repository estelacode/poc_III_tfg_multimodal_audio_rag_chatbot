```python
import gradio as gr
import whisper
import librosa
```

```python
def speech_to_text(audio_file):
    
  # Loading the audio file
  audio, rate = librosa.load(audio_file, sr = 16000)
  # printing audio
  print(audio)
  # printing rate
  print(rate)

  model = whisper.load_model("base")
  result = model.transcribe(audio, language = "English")
  return result["text"]

```

```python
speech_to_text("../data/audio/red_sandals.wav") 
```

```python
demo = gr.Interface(
    fn = speech_to_text,
    inputs = gr.Audio(
        sources = ["microphone"],
        format= "wav",
        type="filepath"
    ),
    outputs = gr.Textbox(),
    title = "Speech to text",
    description = "Speech recognition, also known as automatic speech recognition (ASR), computer speech recognition or speech-to-text, is a capability that enables a program to process human speech into a written format."
)
```

```python
demo.launch()
```
