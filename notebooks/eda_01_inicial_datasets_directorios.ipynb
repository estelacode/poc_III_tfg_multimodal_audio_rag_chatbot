```python
# Ruta del directorio
directorio = '../data/store_zara_data'

# Obtener la lista de archivos en el directorio que terminen con .csv
archivos = [f for f in os.listdir(directorio) if f.endswith('.csv')]

print(archivos)
```

```python
def get_info_datasets(path:str):
    """ Devuelve un DataFrame con la información de los csvs contenido en cada direcctorio """
    #csvs = os.listdir(path)
    csvs = [f for f in os.listdir(path) if f.endswith('.csv')]
    csvs_content = []

    for csv in csvs:
        df = pd.read_csv(path +'/'+ csv, sep=",")
        csvs_content.append([csv, df.shape[0], df.shape[1], df.columns])

    df_csvs = pd.DataFrame(csvs_content, columns=["csv", "filas", "columnas", "nombres_columnas"])
    df = df_csvs.sort_values("filas", ascending=False)
    return df
```
