## EDA 2: Análisis Exploratorio de los datos

* `Directorio:` ../data/store_zara_data
* `csv:`store_zara.csv

```python
import pandas as pd
from PIL import Image
import matplotlib.pyplot as plt
import json
```

```python
# Cargar csv
df_store_zara = pd.read_csv('../data/store_zara_data/store_zara.csv', sep=',')
df_store_zara.head(2)
```

```python
# Mostrar primera fila del dataframe
df_store_zara.loc[0, :]
```

```python
# mostrar primera fila del dataframe de la columna image_downloads
cadena = df_store_zara.loc[0, 'image_downloads']
print(type(cadena))
print(cadena)
json_cadena = cadena.replace("'", "\"")
print(type(json_cadena))
print(json_cadena)
# Parsear el arreglo JSON
lista_imagenes = json.loads(json_cadena)

# lista_imgenes
print(type(lista_imagenes))
print(lista_imagenes)
```

```python
lista_imagenes[0]
```

```python
image_path = "../data/store_zara_data/images/zara/" + str(lista_imagenes[0]) + ".jpg"
print(image_path)
image = Image.open(image_path).convert("RGB")
plt.figure(figsize=(16, 5))
plt.imshow(image)
```

```python
for image in lista_imagenes:
    image_path = "../data/store_zara_data/images/zara/" + str(image) + ".jpg"
    image = Image.open(image_path).convert("RGB")
    plt.figure(figsize=(16, 5))
    plt.imshow(image)
    plt.show()
```
