## EDA 3: Análisis Bivariante

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_store_zara = pd.read_csv('../data/store_zara_data/store_zara.csv', sep=',')
df_store_zara.head(2)
```

```python
df_zara = df_store_zara.loc[:, ['section', 'terms', 'price','currency']]
df_zara.head(2)
```

```python
def get_bivariate_analysis(data):
    """
    Realizar un análsis bivariante de todas las variables de interés con respecto a la variable género.
    """
    column_names = data.columns.tolist()
    custom_palette = {'MAN': '#1079E6', 'WOMAN': '#E649AE'}  # Customize colors for 'M' (Male) and 'F' (Female)
    
    for column in column_names:
        fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(20,5))
        ax = sns.countplot(data=data, x=column, hue = 'section',  ax=ax, legend=True, palette=custom_palette)
        ax.set_title(column, fontsize=20)
        ax.set_xlabel(column,fontsize=0)
        ax.set_ylabel('count', fontsize=10)
        ax.tick_params(labelsize=10)
        ax.tick_params(rotation=90)

        print(f'Tabla de frecuencia: {column}')
        print('*********************************')
        # Obtener Tabla de frecuencias para cada una de las colmunas seleccionadas y adquirir una vision detallada de los datos de cada variable.
        print(data[column].value_counts())
        plt.show()

get_bivariate_analysis(df_zara)
```
